# Q6. The duplicate that isn't

     ### Prediction

I predict that `df.drop_duplicates()` will remove one of the two A-1001 rows because they are exact duplicates. I predict that both A-1002 rows will remain because their seats, MRR, and last_active values are different.

````python
print("Rows before drop_duplicates():", len(df))

deduped = df.drop_duplicates()

print("Rows after drop_duplicates():", len(deduped))
````

````python
print("A-1001 rows before:")
print(df[df['account_id'] == 'A-1001'])

print("\nA-1001 rows after:")
print(deduped[deduped['account_id'] == 'A-1001'])

print("\nA-1002 rows before:")
print(df[df['account_id'] == 'A-1002'])

print("\nA-1002 rows after:")
print(deduped[deduped['account_id'] == 'A-1002'])
````

### Actual result

`df.drop_duplicates()` reduced the dataset from 17 rows to 16 rows.

It removed one of the two A-1001 records because those rows are exact duplicates. Both A-1002 records remained because they are not identical: the seats, MRR, and last_active values differ.

### Why drop_duplicates() is the wrong tool for A-1002

`drop_duplicates()` compares complete rows. It does not decide whether two records with the same account_id represent duplicate data or legitimate historical updates.

A-1002 appears twice, but the second record changes from 5 seats and 450 MRR to 6 seats and 540 MRR, with a later last_active date. Therefore, I would not delete one of the A-1002 records without understanding what one row represents.

### Replacement approach

If the business requirement is to keep only the latest record for each account, I would first identify the correct date field and then use:

```python

    errors='coerce'

    df.sort_values('last_active_dt')
      .drop_duplicates(
          subset='account_id',

**A-1005:** The value is -99/8 = -12.375 MRR per seat. The most likely explanation is that -99 is a special code rather than an actual negative MRR value. I would investigate the source-system definition and treat it accordingly rather than interpreting it as negative revenue.

**A-1012:** The row has 1,000,000 seats and 90 MRR, producing 0.00009 MRR per seat. This is highly suspicious. There are two competing explanations: the seat count could be wrong, or the MRR could be wrong. The data alone cannot distinguish these explanations. **A-1007:** The value is 120 MRR per seat, which differs from the dominant 90 baseline but is not automatically evidence of bad data. I would not change it without additional business evidence.

### Q8 Explanation

A parsing problem occurs when the raw value cannot be interpreted using the expected data representation. A-1003 is an example because signup_ts contains 1675209600, which is not being interpreted as an ordinary date string by the parsing operation.

A validation problem occurs when a value can be successfully parsed but violates a business or logical rule. A-1007 has a valid date in last_active, but its last_active date is before its signup date. A-1013 is another example: 2099-01-01 is a valid date format, but it is an impossible future activity date in the context of this dataset.

# Q9. The same null, two fates

````python
print("Missing company:")
print(
    df[
        df['company'].isna()
    ]
)

print("\nMissing seats:")
print(
    df[
        df['seats'].isna()
    ]
)
````

### Q9 Explanation

A-1014 is missing a value in the label column `company`. A-1008 is missing a value in the measure column `seats`.

The same action, such as dropping the row, can have different consequences depending on the downstream analysis. If I were grouping MRR by company, A-1014 cannot be assigned to a company group, so excluding it from that particular analysis could be appropriate if the company cannot be recovered.

A-1008 is different because its MRR value is still present. If the downstream use is calculating total MRR, dropping A-1008 would unnecessarily remove valid revenue information. However, if the downstream use is calculating MRR per seat, A-1008 cannot be used until the missing seats value is resolved.

# Q10. Cleaning is a function of the question

### Decision 1: Removing the exact A-1001 duplicate

Earlier, I identified the repeated A-1001 row as an exact duplicate and would remove one copy when creating a clean account-level dataset. However, if the goal were to investigate source-system duplication or data-ingestion errors, I would keep both records because the duplicate itself would be useful evidence of the problem.

### Decision 2: Handling A-1008's missing seats

For an MRR-per-seat analysis, I would exclude A-1008 from the calculation until its missing seats value can be resolved because the ratio cannot be calculated. However, if the goal were to calculate total MRR, I would keep A-1008 because its MRR value is still valid and useful.

### Conclusion

"Clean" is not a permanent property of a dataset; it depends on the question the data is being used to answer. A record or value that should be removed or excluded for one analysis can be useful and should be retained for another analysis.
